```python
#BTVN thú vị 
# cho ma trận = [[99,99,99],
#                [99,99,99],
#                [99,99,99]]
# hãy tưởng tượng ô 99 trống
#giả sử 1 là x
#giả sử 0 là o

#lấy đầu vào từ phía x và phía 0 luân phiên
# Nếu đầu vào từ phía x là (0,0)

#ma trân sẽ trở thành
# [[99,99,99],
# [99,99,99],
#[99,99,99]]

#nếu đầu vào từ phía o cũng là (0,0)
#thông báo cho họ nhập lại
#nếu không thì điền vào ma trận

#thử thách: nếu người đầu tiên có 3 phần tử liên tiếp thì dừng trò chơi
```

```python
import numpy as np

board = np.full((3, 3), 99)
def print_board():
    print(board)
    print()
def check_win(player):
    # kiểm tra hàng
    for i in range(3):
        if np.all(board[i, :] == player):
            return True

    # kiểm tra cột
    for i in range(3):
        if np.all(board[:, i] == player):
            return True

    # kiểm tra chéo
    if np.all(np.diag(board) == player):
        return True

    if np.all(np.diag(np.fliplr(board)) == player):
        return True

    return False
current_player = 1  

while True:
    print_board()

    if current_player == 1:
        print("Lượt của X")
    else:
        print("Lượt của O")

    row = int(input("Nhập dòng (0-2): "))
    col = int(input("Nhập cột (0-2): "))

    # kiểm tra ô trống
    if board[row, col] != 99:
        print("Ô này đã được đánh, nhập lại!")
        continue

    # điền vào ma trận
    board[row, col] = current_player

    # kiểm tra thắng
    if check_win(current_player):
        print_board()
        print("X thắng!" if current_player == 1 else "O thắng!")
        break

    # đổi lượt
    current_player = 0 if current_player == 1 else 1

```
